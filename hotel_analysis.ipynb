📊 Exploratory Data Analysis

```python
plt.figure(figsize=(12,8))

sns.heatmap(df[num_cols].corr(),
            annot=True,
            cmap="coolwarm")
plt.savefig('correlation_matrix_heatmap.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
df.hist(figsize=(15,10))
plt.tight_layout()
plt.savefig('Distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
for col in num_cols:
    plt.figure(figsize=(6,3))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.savefig(f'boxplot_{col}.png', dpi=300, bbox_inches='tight')
    plt.show()
```

```python
df["Country"].value_counts().head(10).plot(
    kind="bar",
    figsize=(8,5)
)
plt.title("Top Countries")
plt.xlabel("Country")
plt.ylabel("Count")
plt.savefig('top_countries_bar_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
sns.countplot(data=df,
              x="RoomType")
plt.xticks(rotation=45)
plt.title("Room Type Count")
plt.savefig('room_type_count.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
sns.countplot(data=df,
              x="PaymentStatus")
plt.xticks(rotation=45)
plt.savefig('payment_status_count.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
sns.barplot(
    data=df,
    x="RoomType",
    y="TotalAmount",
    estimator=sum
)
plt.xticks(rotation=45)
plt.title("Total Amount by Room Type")
plt.savefig('total_amount_by_room_type.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
sns.countplot(
    data=df,
    x="BookingChannel"
)
plt.xticks(rotation=45)
plt.savefig('booking_channel_count.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
sns.countplot(
    data=df,
    x="IsCancelled"
)
plt.savefig('cancellation_count.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
df["CheckInDate"] = pd.to_datetime(df["CheckInDate"])
df["CheckOutDate"] = pd.to_datetime(df["CheckOutDate"])
df["StayDays"] = (
    df["CheckOutDate"] -
    df["CheckInDate"]
).dt.days
df.head()
```

```python
sns.histplot(df["StayDays"],
             bins=20)
plt.xlabel("Stay Days")
plt.ylabel("Frequency")
plt.title("Distribution of Stay Days")
plt.savefig('stay_days_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
df.groupby("RoomType")["StayDays"].mean().sort_values()
```

```python
df.groupby("Country")["TotalAmount"]\
.sum()\
.sort_values(ascending=False)\
.head(10)
```

```python
df["Month"] = df["CheckInDate"].dt.to_period("M")
monthly = df.groupby("Month")["TotalAmount"].sum()
monthly.plot(figsize=(12,5))
plt.title("Monthly Total Amount")
plt.xlabel("Month")
plt.ylabel("Total Amount")
plt.savefig('monthly_total_amount.png', dpi=300, bbox_inches='tight')
plt.show()
```
